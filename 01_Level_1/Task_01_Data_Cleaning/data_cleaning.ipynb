```python
import pandas as pd
file_path="D:/Documents/Projects/Intern/Datasets_Provided/Churn Prdiction Data/churn-bigml-80.csv"
df = pd.read_csv(file_path)
df.head()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df_clean = df.copy()
```

```python
missing_values = df_clean.isnull().sum()

missing_values
```

```python
print("Total missing values:", df_clean.isnull().sum().sum())
```

```python
duplicate_count = df_clean.duplicated().sum()

print("Duplicate rows:", duplicate_count)
```

```python
df_clean = df_clean.drop_duplicates()
```

```python
df_clean.dtypes
```

```python
df_clean.info()
```

```python
df_clean.columns.tolist()
```

```python
categorical_columns = [
    "State",
    "International plan",
    "Voice mail plan",
    "Churn"
]

for column in categorical_columns:
    print("\n", column)
    print(df_clean[column].value_counts(dropna=False))
```

```python
for column in categorical_columns:
    print(column, ":", df_clean[column].unique())
```

```python
numeric_columns = df_clean.select_dtypes(include="number").columns

negative_counts = (df_clean[numeric_columns] < 0).sum()
negative_counts[negative_counts > 0]
```

```python
output_path = "cleaned_churn_data.csv"

df_clean.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully.")
```
